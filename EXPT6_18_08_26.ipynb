```python
mean = df["alcohol"].mean()
median = df["alcohol"].median()
mode = df["alcohol"].mode()

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode.values)
```

```python
std = df["alcohol"].std()

print("Standard Deviation:", std)
```

```python
variance = df["alcohol"].var()

print("Variance:", variance)
```

```python
range_value = df["alcohol"].max() - df["alcohol"].min()

print("Range:", range_value)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["alcohol"], bins=20)

plt.title("Distribution of Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["alcohol"])

plt.title("Box Plot of Alcohol")
plt.xlabel("Alcohol")

plt.show()
```

```python
Q1 = df["alcohol"].quantile(0.25)
Q3 = df["alcohol"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["alcohol"] < lower_limit) |
    (df["alcohol"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Number of outliers:", len(outliers))
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["alcohol"], df["quality"])

plt.title("Alcohol vs Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="quality", y="alcohol", data=df)

plt.title("Alcohol Distribution by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["volatile acidity"], df["quality"])

plt.title("Volatile Acidity vs Wine Quality")
plt.xlabel("Volatile Acidity")
plt.ylabel("Quality")

plt.show()
```

```python
selected_columns = [
    "alcohol",
    "volatile acidity",
    "citric acid",
    "pH",
    "quality"
]

sns.pairplot(
    df[selected_columns]
)

plt.show()
```

```python
sns.pairplot(
    df[
        ["alcohol", "volatile acidity", "sulphates", "quality"]
    ],
    kind="reg"
)

plt.show()
```

```python
df["quality_category"] = pd.cut(
    df["quality"],
    bins=[0, 5, 7, 10],
    labels=["Low", "Medium", "High"]
)

sns.pairplot(
    df[
        ["alcohol", "volatile acidity", "pH", "quality_category"]
    ],
    hue="quality_category"
)

plt.show()
```

```python
corr = stats.pearsonr(
    df["alcohol"],
    df["quality"]
)

print("Pearson Correlation:", corr[0])
print("P-value:", corr[1])
```

```python
correlation = df.corr(
    method="pearson",
    numeric_only=True
)

display(correlation)
```

```python
corr_pairs = correlation.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

corr_pairs = corr_pairs.sort_values(
    ascending=False
)

print(corr_pairs.head(10))
```

```python
quality_corr = correlation["quality"].sort_values(
    ascending=False
)

print("Correlation with Wine Quality:")
print(quality_corr)
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Wine Dataset")

plt.show()
```

```python
wine_features = df.drop(
    columns=["Id", "quality_category"],
    errors="ignore"
)

correlation = wine_features.corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Wine Features Correlation Heatmap")

plt.show()
```
