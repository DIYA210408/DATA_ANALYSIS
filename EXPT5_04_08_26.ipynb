```python
bins = [0, 20, 50, 100, 500, 1000, 5000]

labels = [
    "Very Low",
    "Low",
    "Medium",
    "High",
    "Very High",
    "Premium"
]

df["Amount_Category"] = pd.cut(df["amt"],
                               bins=bins,
                               labels=labels)
```

```python
df[["amt","Amount_Category"]].head(20)
```

```python
df["Amount_Category"].value_counts()
```

```python
pd.cut(df["amt"],5)
```

```python
df["Amount_QCut"] = pd.qcut(df["amt"],4)
```

```python
scaler = MinMaxScaler()

df["Amount_MinMax"] = scaler.fit_transform(df[["amt"]])

df[["amt","Amount_MinMax"]].head(10)
```
